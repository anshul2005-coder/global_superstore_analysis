```python
import pandas as pd

# Loads the dataset (Make sure the path matches where you saved your file)
df = pd.read_csv(r'C:\Users\anshu\OneDrive\Desktop\begineer project\Data\Sample - Superstore.csv', encoding='windows-1252') 

# Displays the first 5 rows to make sure it loaded correctly
df.head()
```

```python
# 1. Show column data types and check for missing values
print("--- DATA INFO ---")
df.info()

print("\n--- MISSING VALUES PER COLUMN ---")
# 2. Show the exact count of null spaces
print(df.isnull().sum())
```

```python
# 1. Standardize column names (convert spaces to underscores and make lowercase)
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_').str.lower()

# 2. Converting date-type of columns from 'object' (text) to actual datetime format
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])

# 3. Verifying that the changes worked
print("--- NEW COLUMN NAMES ---")
print(df.columns.tolist())

print("\n--- NEW DATA TYPES ---")
print(df[['order_date', 'ship_date']].dtypes)
```

```python
# Exporting the clean dataframe to a new CSV file
df.to_csv(r'C:\Users\anshu\OneDrive\Desktop\begineer project\Data\clean_superstore.csv', index=False)

print("The clean data has been saved as 'clean_superstore.csv'")
```
